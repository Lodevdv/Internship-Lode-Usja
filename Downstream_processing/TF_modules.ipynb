```python
tf_col = "Unnamed: 0"
df = data["Proliferating_cDC1"]
experiment_cols = df.columns[1:]

other_exps = experiment_cols[experiment_cols != "CITEseq_Toxo_Test"]

df["Toxo_vs_other_mean"] = (
    df["CITEseq_Toxo_Test"] - df[other_exps].mean(axis=1)
)

df_sorted = df.sort_values(
    "Toxo_vs_other_mean",
    ascending=False
)

print(df_sorted[[tf_col, "CITEseq_Toxo_Test", "Toxo_vs_other_mean"]].head(20)["Unnamed: 0"].to_list())
```
